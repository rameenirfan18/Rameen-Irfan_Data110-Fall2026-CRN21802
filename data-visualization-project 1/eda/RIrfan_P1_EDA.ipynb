```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("netflix_titles.csv")
df.head()
```

This folder just contains the data analysis, and extra steps taken before the final result. Examing the data, and making necessary changes

```python
#Figure 2 Exploration:
#Check the genres for next exploration
df["listed_in"].head(10)
```

```python
#Fig 2
#Separating and counting the genres
genres = []

for item in df["listed_in"]:
  genre_list = item.split(",")

  for genre in genre_list:
    genres.append(genre.strip())

genre_counts = pd.Series(genres).value_counts()
genre_counts.head(10)
```

```python
#Figure 3 Data Analysis
#Country exploration
df["country"].head(10)
```

```python
#Fig 3
#Separate and count the countries
countries = []

for item in df["country"]:
  if pd.notna(item):
    country_list = item.split(",")

    for country in country_list:
      countries.append(country.strip())

country_counts = pd.Series(countries).value_counts()
country_counts.head(10)
```

```python
#Figure 4:
#Explore the duration column
df["duration"].head(15)
```

```python
#Fig 4
#Filtering Movies from TV Shows
movies = df[df["type"] == "Movie"]
movies [["title", "duration"]].head(10)
```

```python
#Conversion of movie durations in numbers
movies = movies.copy()
movies["duration_minutes"] = movies["duration"].str.replace(" min", "")
movies["duration_minutes"] = pd.to_numeric(movies["duration_minutes"], errors = "coerce")
movies[["title", "duration_minutes"]].head(10)
```
